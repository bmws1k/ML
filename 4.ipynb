```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder


dataset = pd.read_csv("weather.csv")


encoder = LabelEncoder()

for column in dataset.columns:
    dataset[column] = encoder.fit_transform(dataset[column])


input_data = dataset.iloc[:, :-1]
target_output = dataset.iloc[:, -1]


decision_tree_model = DecisionTreeClassifier()


decision_tree_model.fit(input_data, target_output)

prediction = decision_tree_model.predict([input_data.iloc[2]])
actual = target_output[2]
print("Prediction:", encoder.inverse_transform([prediction]))
print("Actual:", encoder.inverse_transform([actual]))
```
