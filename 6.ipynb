```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,recall_score,precision_score

dataset =pd.read_csv("doc.csv")

document = dataset["Text"]
label = dataset["Label"]
vectorizer = CountVectorizer()

document_mat = vectorizer.fit_transform(document)
x_train,x_test,y_train,y_test=train_test_split(document_mat,label,test_size=0.2,random_state = 0)
model = MultinomialNB()
model.fit(x_train,y_train)
prediction = model.predict(x_test)
print("Prediction",prediction)
print("Actual",list(y_test))
acc=accuracy_score(y_test,prediction)
rec=recall_score(y_test,prediction)
pre=precision_score(y_test,prediction)

print(acc)
print(rec)
print(pre)
```
