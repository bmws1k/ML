```python
import pandas as pd

dataset = pd.read_csv("data2.csv")

concepts = dataset.iloc[:, :-1].values
targets = dataset.iloc[:, -1].values

specific_hypothesis = concepts[0].copy()

general_hypothesis = [["?" for i in range(len(specific_hypothesis))]
                      for i in range(len(specific_hypothesis))]

for i in range(len(targets)):
    if targets[i] == "Yes":
        specific_hypothesis = concepts[i].copy()

for i in range(len(concepts)):

    if targets[i] == "Yes":
        for j in range(len(specific_hypothesis)):
            if concepts[i][j] != specific_hypothesis[j]:
                specific_hypothesis[j] = "?"
                general_hypothesis[j][j] = "?"

    else:
        for j in range(len(specific_hypothesis)):
            if concepts[i][j] != specific_hypothesis[j]:
                general_hypothesis[j][j] = specific_hypothesis[j]
            else:
                general_hypothesis[j][j] = "?"

print("Specific Hypothesis:")
print(specific_hypothesis)

print("\nGeneral Hypothesis:")
for rule in general_hypothesis:
    print(rule)
```
